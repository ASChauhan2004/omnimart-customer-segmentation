### 1. Objective

The objective of this notebook is to understand the structure,
quality, and characteristics of the raw OmniRetail customer dataset
before performing any data-cleaning or analytical transformations.

```python
import pandas as pd
import numpy as np

file_path = "../data/OmniRetail_Customer_Analytics_Group_Project.xlsx" 

df = pd.read_excel(file_path)

print("Shape of the dataset:", df.shape)
display(df.head())
display(df.tail())
```

```python
df.info()
print(df.dtypes)
display(df.describe(include='all').T)
```

## Checking for missing values


```python

missing = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing Percentage": (df.isna().sum() / len(df)) * 100
})

display(
    missing.sort_values("Missing Percentage", ascending=False)
)
```

## Checking the catergorical columns and their unique values


```python

categorical_cols = ["Region", "City_Tier", "Membership_Tier", "Preferred_Category","Acquisition_Channel"]

for col in categorical_cols:
    print("\n===========", col , "===========")
    print(df[col].value_counts(dropna=False))

for col in categorical_cols:
    print("\n", col)
    for value in df[col].dropna().unique():
        print(repr(value))
```

## Checking for duplicate customer IDs


```python

print("Total rows:", len(df))
print("Unique customers:", df['Customer_ID'].nunique())
print(
    "Duplicate customer rows:", df["Customer_ID"].duplicated(keep=False).sum()
)

duplicate_counts = df["Customer_ID"].value_counts()

display(
    duplicate_counts[duplicate_counts > 1].sort_values(ascending=False)
)
```

The data definetely has duplicate customer IDs but we need to also check that whether these rows have only customer ID as same or the whole row is identical. Identical rows can be dropped easily, but, with rows with different non-ID fields that are different have some value to them

```python
duplicate_ids = duplicate_counts[
    duplicate_counts > 1
]

duplicate_rows = df[
    df["Customer_ID"].isin(duplicate_ids.index)
].sort_values("Customer_ID")

duplicate_summary = (
    duplicate_rows
    .groupby("Customer_ID")
    .agg(
        Rows=("Customer_ID", "size"),
        Distinct_Values=("Customer_ID", "nunique")
    )
)

duplicate_columns = [
    col for col in df.columns
    if col != "Customer_ID"
]

duplicate_comparison = (
    duplicate_rows
    .groupby("Customer_ID")[duplicate_columns]
    .nunique(dropna=False)
)

print("Duplicate customer IDs:", len(duplicate_comparison))

print(
    "IDs with completely identical non-ID fields:",
    (duplicate_comparison.max(axis=1) == 1).sum()
)

print(
    "IDs with at least one differing field:",
    (duplicate_comparison.max(axis=1) > 1).sum()
)
```

### Duplicate Customer Finding

The dataset contains 250 Customer_IDs that occur more than once.

Of these:
- 150 Customer_IDs (60%) have completely identical non-ID fields.
- 100 Customer_IDs (40%) have at least one differing field.

The 150 identical duplicates appear to be repeated copies of the
same customer records. The 100 conflicting duplicates require
further investigation before deciding how they should be treated.

```python
conflicting_ids = duplicate_comparison[
    duplicate_comparison.max(axis=1) > 1
].index

conflicting_rows = duplicate_rows[
    duplicate_rows["Customer_ID"].isin(conflicting_ids)
].sort_values("Customer_ID")

difference_summary = {}

for col in duplicate_columns:
    difference_summary[col] = (
        duplicate_comparison.loc[conflicting_ids, col] > 1
    ).sum()

difference_summary = (
    pd.Series(difference_summary)
    .sort_values(ascending=False)
)

display(difference_summary)
```

```python
sample_conflicting_ids = (
    pd.Series(conflicting_ids)
    .sample(5, random_state=42)
    .tolist()
)

display(
    conflicting_rows[
        conflicting_rows["Customer_ID"].isin(sample_conflicting_ids)
    ]
)
```

### Duplicate Variation Finding

Among the 100 Customer_IDs with differing duplicate records, the
variation is concentrated in two variables:

- `Annual_Spend_000` differs for 99 of the 100 duplicated customers.
- `Satisfaction_Score` differs for 89 of the 100 duplicated customers.

All other inspected variables, including demographic, geographic,
membership, behavioral, and date fields, are identical across the
duplicate records.

This suggests that the repeated records may represent duplicated
customer observations with altered spend and satisfaction values.
The records will therefore be investigated further before any
deduplication decision is made.

```python
comparison_stats = (
    conflicting_rows
    .groupby("Customer_ID")
    .agg(
        Spend_Min=("Annual_Spend_000", "min"),
        Spend_Max=("Annual_Spend_000", "max"),
        Satisfaction_Min=("Satisfaction_Score", "min"),
        Satisfaction_Max=("Satisfaction_Score", "max")
    )
)

comparison_stats["Spend_Difference"] = (
    comparison_stats["Spend_Max"]
    - comparison_stats["Spend_Min"]
)

comparison_stats["Satisfaction_Difference"] = (
    comparison_stats["Satisfaction_Max"]
    - comparison_stats["Satisfaction_Min"]
)

display(comparison_stats.describe())

sample_conflicting_ids = (
    pd.Series(conflicting_ids)
    .sample(5, random_state=42)
    .tolist()
)

sample_duplicates = conflicting_rows[
    conflicting_rows["Customer_ID"].isin(sample_conflicting_ids)
].sort_values("Customer_ID")

display(
    sample_duplicates[
        [
            "Customer_ID",
            "Annual_Spend_000",
            "Satisfaction_Score"
        ]
    ]
)
```

### Conflicting Duplicate Inspection

Inspection of sample records shows that the 100 conflicting
Customer_IDs have identical values across all fields except
`Annual_Spend_000` and, in most cases, `Satisfaction_Score`.

The differences observed in the sample are generally relatively small,
but the repeated records cannot be treated as exact duplicates because
their customer-level values are not identical.

These records are therefore flagged for explicit treatment during the
cleaning stage rather than being removed automatically.

## Analysing the distribution of numerical columns

## Inspecting the dates columns


```python

date_cols = ["Snapshot_Date", "Last_Purchase_Date", "Signup_Date"]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

for col in date_cols:
    print("\n", col)
    print("Min:", df[col].min())
    print("Max:", df[col].max())
    print("Missing:", df[col].isna().sum())
    print("Unique:", df[col].nunique())

display(df[date_cols].head(10))

#checking for failed conversions if any
for col in date_cols:
    print(
        col,
        "invalid/unparseable dates:",
        df[col].isna().sum()
    )

#checking for impossible dates (last purchase date before signup date)
print(
    "Signup after snapshot:",
    (df["Signup_Date"] > df["Snapshot_Date"]).sum()
)

print(
    "Last purchase after snapshot:",
    (df["Last_Purchase_Date"] > df["Snapshot_Date"]).sum()
)

print(
    "Last purchase before signup:",
    (df["Last_Purchase_Date"] < df["Signup_Date"]).sum()
)
```
